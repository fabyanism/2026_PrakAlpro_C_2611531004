## 1.1 Latar Belakang
Seiring dengan perkembangan teknologi informasi yang pesat, penggunaan sistem terkomputerisasi dalam kegiatan transaksi jual beli telah menjadi kebutuhan penting di berbagai sektor usaha. Proses transaksi secara manual sering kali memakan waktu lebih lama dan memiliki risiko kesalahan perhitungan (human error) yang dapat merugikan pihak penjual maupun pembeli. Oleh karena itu, penerapakan program kasir yang mampu melakukan perhitungan otomatis sangat diperlukan untuk meningkatkan efisiensi dan akurasi pembayaran.Dalam dunia pemrograman, bahasa Python menjadi salah satu pilihan utama untuk membangun aplikasi sederhana hingga kompleks karena sintaksnya yang mudah dipahami, efisien, dan fleksibel. Salah satu fitur dasar namun krusial dalam pemrograman Python adalah struktur kontrol percabangan (if-else). Fitur ini memungkinkan program untuk mengambil keputusan secara otomatis berdasarkan kondisi tertentu, seperti memberikan potongan harga (diskon) apabila total belanjaan pelanggan memenuhi kriteria minimal transaksi.Berdasarkan permasalahan dan konsep tersebut, praktikum ini dilaksanakan untuk merancang dan mengimplementasikan sebuah Program Kasir dan Perhitungan Diskon Otomatis berbasis Python. Program ini dibuat tanpa modul tambahan untuk melatih pemahaman mendalam mengenai logika dasar komputer, pemrosesan variabel masukan (input), serta penerapan logika kondisi (if-else) dalam menyelesaikan kasus transaksi dunia nyata secara cepat dan tepat.

## 2.1 Dasar Teori
Tuliskan penjelasan konsep/teori yang mendasari praktikum ini (definisi, elemen-elemen penting, dsb). Bisa memakai poin bernomor untuk merinci sub-konsep, contoh:

1. Bahasa Pemrograman Python
Python merupakan bahasa pemrograman tingkat tinggi yang bersifat interpreted dan berfokus pada keterbacaan kode. Bahasa ini cocok untuk pemula karena struktur sintaksnya intuitif dan tidak memerlukan deklarasi variabel yang rumit.
2. Fungsi Masukan dan Keluaran (input & print)
Fungsi input() digunakan untuk menerima masukan data dari pengguna berupa teks (string). Fungsi print() berfungsi untuk menampilkan teks, variabel, atau hasil eksekusi program ke layar.
3. Tipe Data dan Konversi Tipe Data (Casting): Secara default, masukan dari input() bertipe string. Untuk melakukan operasi aritmatika, data harus dikonversi (casting) ke tipe numerik seperti float (bilangan desimal) atau int (bilangan bulat).
4. Struktur Percabangan (if-else): Percabangan digunakan untuk mengontrol alur program berdasarkan kondisi tertentu. Jika kondisi pada if bernilai True (benar), maka blok kode if yang dijalankan. Jika bernilai False (salah), program akan menjalankan blok kode else.
5. Operator Aritmatika dan Perbandingan: Operator perbandingan (seperti >) digunakan untuk mengevaluasi kondisi. Operator aritmatika (seperti * dan -) digunakan untuk menghitung nilai matematika, seperti kalkulasi diskon dan total bayar.
