## 1.1 Latar Belakang
Setelah mengenal dasar penulisan program dan perangkat pendukungnya pada pekan sebelumnya, tahap berikutnya dalam mempelajari pemrograman adalah memahami bagaimana sebuah program menyimpan dan mengolah data. Setiap data yang diproses komputer memiliki tipe tertentu, dan Python sebagai bahasa dengan dynamic typing menentukan tipe data secara otomatis berdasarkan nilai yang diberikan kepada sebuah variabel. Pemahaman terhadap tipe data menjadi penting karena kesalahan dalam menentukan tipe sering menjadi penyebab program gagal berjalan, misalnya ketika masukan pengguna yang sebenarnya bertipe string langsung dibandingkan dengan bilangan.

Di sisi lain, sebuah program tidak dapat berdiri sendiri tanpa interaksi dengan penggunanya. Proses masukan dan keluaran (input/output) melalui fungsi input() dan print() menjadi jembatan antara pengguna dan program. Karena fungsi input() selalu mengembalikan nilai bertipe string, diperlukan proses konversi tipe data (type casting) agar nilai tersebut dapat diolah secara numerik. Selain itu, terdapat nilai-nilai tertentu yang tidak boleh berubah selama program berjalan, seperti batas nilai kelulusan, sehingga diperlukan penerapan konstanta menggunakan modul typing.Final. Oleh karena itu, praktikum pekan ini membahas tipe data, variabel, konstanta, konversi tipe, serta operasi masukan dan keluaran secara terpadu melalui studi kasus sistem registrasi praktikan.

## 2.1 Dasar Teori

1. Variabel dan Aturan Penamaan
Variabel merupakan nama yang digunakan untuk menyimpan nilai di dalam memori komputer. Python menganut sifat case-sensitive, sehingga Nama_1004 dan nama_1004 dianggap sebagai dua variabel yang berbeda. Nama variabel tidak boleh diawali angka, tidak boleh mengandung spasi, dan tidak boleh menggunakan kata kunci bawaan (reserved keywords) seperti if, for, atau class. Penamaan variabel dianjurkan menggunakan huruf kecil dengan pemisah garis bawah (snake case) agar kode mudah dibaca.
2. Tipe Data Dasar Python
Python menyediakan beberapa tipe data bawaan. Tipe str digunakan untuk menyimpan teks yang ditulis di antara tanda petik tunggal, ganda, maupun petik tiga ("""...""") untuk teks multiline. Tipe int menyimpan bilangan bulat, float menyimpan bilangan desimal, dan complex menyimpan bilangan kompleks yang terdiri atas bagian riil dan imajiner dengan notasi j (contoh: 100+3j). Perlu dicatat bahwa Python tidak memiliki tipe data char tersendiri seperti pada bahasa C, sehingga karakter tunggal tetap direpresentasikan sebagai str dengan panjang satu.
3. Tipe Data Boolean dan Operator Perbandingan
Tipe bool hanya memiliki dua nilai, yaitu True dan False. Nilai boolean umumnya dihasilkan dari operasi perbandingan menggunakan operator seperti >, <, >=, <=, ==, dan !=. Sebagai contoh, ekspresi nilai_1004 >= BATAS_LULUS_1004 akan menghasilkan True apabila skor yang dimasukkan lebih besar atau sama dengan batas kelulusan, dan False apabila sebaliknya.
4. Konstanta dengan typing.Final
Python tidak menyediakan kata kunci khusus untuk konstanta. Sebagai gantinya, modul typing menyediakan anotasi Final yang menandakan bahwa sebuah nama tidak dimaksudkan untuk diubah nilainya setelah diinisialisasi. Konstanta dideklarasikan di awal program dan ditulis menggunakan huruf kapital seluruhnya, misalnya BATAS_LULUS_1004: Final = 75.0. Anotasi ini bersifat penanda bagi type checker dan pembaca kode, bukan penguncian nilai secara paksa oleh interpreter.
5. Input, Output, dan Type Casting
Fungsi print() digunakan untuk menampilkan keluaran ke layar, sedangkan fungsi input() digunakan untuk menerima masukan dari pengguna melalui konsol. Nilai yang dikembalikan input() selalu bertipe str, sehingga apabila masukan tersebut akan diolah secara aritmetika perlu dilakukan konversi tipe data menggunakan fungsi int() atau float(). Proses perubahan tipe data ini disebut type casting.
6. Fungsi type()
Fungsi bawaan type() digunakan untuk mengetahui tipe data dari sebuah nilai atau variabel. Keluaran fungsi ini berupa <class 'tipe'>, misalnya <class 'int'> untuk bilangan bulat. Fungsi ini bermanfaat untuk memverifikasi bahwa proses konversi tipe data telah berjalan sesuai harapan.

### 2.2.1 Program "Numerik_2611531004"
Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan (tidak perlu screenshot terpisah untuk hasil run).

```python
# Program pertama
"""
Komentar
dua baris atau lebih
pertemuan 1 praktikum alpro
"""
print("Hello World")
print()
print("Departemen Informatika")
print("UNiversitas Andalas")
print("/ \\ // \\\\ /// \\\\\\")
print("program ini" + " dibuat oleh a\\n " + "Fabian Alhadi Zalinov dan 2611531004.")
```

**Penjelasan :**
1. Bagian pertama (baris komentar): Baris """Komentar dua baris atau lebih pertemuan 1 praktikum alpro""" merupakan komentar banyak baris (multi-line comment) yang diapit oleh tiga tanda kutip ("""..."""). Bagian ini tidak dieksekusi oleh Python dan berfungsi sebagai dokumentasi awal program, menjelaskan bahwa kode ini dibuat untuk pertemuan 1 praktikum Algoritma Pemrograman.
2. 3. Bagian ketiga (dekorasi dan penggabungan string): Baris print("/ \\ // \\\\ /// \\\\\\") menampilkan karakter garis miring sebagai elemen dekoratif/hiasan pada output, dengan tanda \\ digunakan sebagai karakter escape untuk menampilkan tanda backslash (\) secara harfiah. Baris terakhir print("program ini" + " dibuat oleh a\\n " + "Fabian Alhadi Zalinov dan 2611531004.") menggabungkan (concatenate) tiga potongan string menggunakan operator + menjadi satu kalimat utuh, menampilkan keterangan pembuat program. Perlu dicatat bahwa \\n di sini ditulis dengan escape ganda sehingga akan tampil sebagai teks \n biasa, bukan sebagai baris baru (newline).
