## 2.1 Dasar Teori

1. Perulangan for dan fungsi range(): Perulangan for pada Python digunakan untuk mengiterasi elemen dari suatu urutan (sequence) atau objek yang dapat diiterasi. Untuk perulangan dengan jumlah iterasi tertentu, perulangan ini lazim dipasangkan dengan fungsi range(awal, akhir, langkah). Fungsi tersebut menghasilkan deret bilangan bulat dari nilai awal hingga satu sebelum nilai akhir. Nilai langkah negatif memungkinkan iterasi menurun, misalnya range(4, 0, -1) yang menghasilkan 4, 3, 2, 1.
2. Perulangan bersarang (nested loop): Perulangan bersarang adalah perulangan yang berada di dalam badan perulangan lain. Perulangan luar mengatur baris, sedangkan perulangan dalam mengatur isi setiap baris. Setiap satu iterasi perulangan luar akan menjalankan seluruh iterasi perulangan dalam hingga selesai. Struktur ini sangat sesuai untuk membentuk pola dua dimensi, karena jumlah spasi dan deret angka pada tiap baris dapat bergantung pada variabel baris saat itu.
3. Fungsi print() dengan parameter end, input(), dan int(): Secara bawaan, print() mengakhiri keluaran dengan karakter baris baru. Parameter end="" mengubah perilaku ini sehingga keluaran berikutnya tetap berada pada baris yang sama. Hal ini penting untuk menyusun satu baris pola dari banyak potongan karakter. Pemanggilan print() tanpa argumen di akhir baris digunakan untuk berpindah ke baris berikutnya. Masukan dari pengguna dibaca dengan input() sebagai tipe string, sehingga perlu dikonversi dengan int() agar dapat dipakai dalam perhitungan.
4. Formula matematis pembentukan pola: Pola jam pasir dibentuk dari hubungan aritmetika antara ukuran N dan nomor baris. Pada studi kasus ini, lebar bingkai horizontal adalah 4 * N + 5 karakter =, sedangkan spasi penyeimbang pada baris ke-b sebanyak 2 * (N - b). Spasi penyeimbang pada fase poros pusat sebanyak 2 * N + 1. Karena perkalian string tidak boleh dipakai, nilai-nilai tersebut menjadi batas perulangan for yang mencetak satu karakter pada setiap iterasi.
5. Git dan GitHub: Git adalah sistem version control terdistribusi untuk melacak perubahan pada kode program, sedangkan GitHub adalah platform daring yang menyimpan repositori Git dan menjadi pusat kolaborasi pengembang. Proses pengunggahan berkas meliputi staging (git add), pencatatan perubahan (git commit), dan pengunggahan ke repositori jarak jauh (git push).

### 2.2.3 Program "perulangan_for3_2611531004.py"
Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan (tidak perlu screenshot terpisah untuk hasil run).

```python
# Buat file dengan nama perulangan_for3_2611531004.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1004
# Program ini menggunakan fungsi input()

ulang_1004 = int(input("Masukkan jumlah perulangan: "))

jumlah_1004 = 0
for i_1004 in range(1, ulang_1004 + 1):
    print(i_1004, end=" ")
    jumlah_1004 = jumlah_1004 + i_1004

    if i_1004 < ulang_1004:
        print(" + ", end=" ")
    else:
        print(" = ", jumlah_1004, end=" ")
print()
print("Jumlah =", jumlah_1004)
```

### 2.2.8 Program "nested_for4_2611531004.py"
Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan (tidak perlu screenshot terpisah untuk hasil run).

```python
# Buat file dengan nama nested_for4_2611531004.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1004
# Program ini menggunakan fungsi input()

tinggi_1004 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_1004 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_1004 = tinggi_1004
    c_1004 = a_1004
    lebar_1004 = (2 * tinggi_1004) - 2

    for i_1004 in range(1, tinggi_1004 + 1):
        b_1004 = c_1004 + 1

        for j_1004 in range(1, lebar_1004 + 1):

            # Baris atas atau bawah
            if i_1004 == 1 or i_1004 == tinggi_1004:
                if j_1004 == 1 or j_1004 == lebar_1004:
                    print("#", end="")
                else:
                    print("=", end="")

            # Baris isi
            else:
                if j_1004 == 1 or j_1004 == lebar_1004:
                    print("|", end="")
                else:
                    if j_1004 == c_1004:
                        print("<", end="")
                    elif j_1004 == b_1004:
                        print(">", end="")
                    elif j_1004 == (lebar_1004 - c_1004):
                        print("<", end="")
                    elif j_1004 == (lebar_1004 - c_1004 + 1):
                        print(">", end="")
                    elif j_1004 > b_1004 and j_1004 < (lebar_1004 - c_1004):
                        print("-", end="")
                    else:
                        print(" ", end="")

        print()

        # Logika asli Java
        a_1004 -= 2

        if a_1004 <= 0:
            c_1004 = (-a_1004) + 2
        else:
            c_1004 = a_1004
```

**Penjelasan :**
1. Bagian pertama kode ini berfungsi untuk meminta tinggi pola dari pengguna dan menyimpannya di tinggi_1004.
2. Bagian kedua berfungsi untuk memvalidasi input dengan if tinggi_1004 % 2 != 0. Jika tinggi ganjil, program menampilkan pesan "Tinggi harus bilangan genap!" dan berhenti. Jika genap, program lanjut ke pembuatan pola.
3. Bagian ketiga berfungsi untuk menyiapkan variabel awal. a_1004 dan c_1004 diisi tinggi_1004, dan lebar_1004 dihitung dengan rumus (2 * tinggi_1004) - 2. c_1004 menjadi penentu posisi tanda "<" di tiap baris, sedangkan a_1004 pengaturnya.
4. Bagian keempat berfungsi sebagai perulangan luar yang menjalankan baris dari 1 sampai tinggi_1004. Di awal setiap baris, b_1004 dihitung sebagai c_1004 + 1, yaitu posisi tanda ">" pasangan dari "<".
5. Bagian kelima berfungsi sebagai perulangan dalam yang berjalan dari kolom 1 sampai lebar_1004, mencetak satu karakter per kolom.
6. Bagian keenam berfungsi untuk menggambar baris paling atas dan paling bawah. Kolom pertama dan terakhir dicetak "#", sedangkan kolom di antaranya dicetak "=".
7. Bagian ketujuh berfungsi untuk menggambar baris isi (di antara baris atas dan bawah). Kolom paling kiri dan kanan dicetak "|".
8. Bagian kedelapan berfungsi untuk menentukan karakter di dalam baris isi. Kolom c_1004 dan lebar_1004 - c_1004 dicetak "<", sedangkan kolom b_1004 dan lebar_1004 - c_1004 + 1 dicetak ">". Kolom di antara pasangan tanda tersebut diisi "-", dan sisanya diisi spasi kosong.
9. Bagian kesembilan berfungsi untuk pindah baris dengan print() setelah satu baris penuh dicetak.
10. Bagian kesepuluh berfungsi untuk memperbarui posisi tanda di baris berikutnya. a_1004 dikurangi 2. Jika a_1004 masih positif, c_1004 mengikuti nilainya, yaitu tanda bergerak ke arah tepi. Jika a_1004 sudah nol atau negatif, c_1004 = -a_1004 + 2, yaitu tanda bergerak kembali ke tengah. Inilah yang membuat pola melebar lalu menyempit secara simetris.

### 2.2.9 Program "latihan5_2611531004.py"
Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan (tidak perlu screenshot terpisah untuk hasil run).

```python
# Buat file dengan nama latihan5_2611531004.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1004
# Program ini menggunakan fungsi input()

tinggi_1004 = int(input("Masukkan tinggi segitiga: "))

for i_1004 in range(1, tinggi_1004 + 1):
    print(" " * (tinggi_1004 - i_1004), end="")
    print("* " * i_1004)
```

```python
# Nama berkas : tugas5_2611531004.py
# Program     : Pola Jam Pasir Kristal Palindromik Berbingkai
# Catatan     : Seluruh border, spasi, dan deret angka dibentuk dengan
#               perulangan for bersarang (print dengan end="").

print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")
n_1004 = int(input("Masukkan ukuran skala jam pasir (N): "))

# ---------- Bingkai atas ----------
print("#", end="")
for garis_1004 in range(4 * n_1004 + 5):
    print("=", end="")
print("#")

# ---------- Fase 1: Jam pasir atas (baris N turun s.d. 1) ----------
for baris_1004 in range(n_1004, 0, -1):
    print("| ", end="")
    for spasi_1004 in range(2 * (n_1004 - baris_1004)):
        print(" ", end="")
    for angka_1004 in range(baris_1004, 0, -1):
        print(angka_1004, end=" ")
    print("<*>", end="")
    for angka_1004 in range(1, baris_1004 + 1):
        print(" ", end="")
        print(angka_1004, end="")
    for spasi_1004 in range(2 * (n_1004 - baris_1004)):
        print(" ", end="")
    print(" |")

# ---------- Fase 2: Poros titik pusat ----------
print("|", end="")
for spasi_1004 in range(2 * n_1004 + 1):
    print(" ", end="")
print("<*>", end="")
for spasi_1004 in range(2 * n_1004 + 1):
    print(" ", end="")
print("|")

# ---------- Fase 3: Jam pasir bawah (baris 1 naik s.d. N) ----------
for baris_1004 in range(1, n_1004 + 1):
    print("| ", end="")
    for spasi_1004 in range(2 * (n_1004 - baris_1004)):
        print(" ", end="")
    for angka_1004 in range(baris_1004, 0, -1):
        print(angka_1004, end=" ")
    print("<*>", end="")
    for angka_1004 in range(1, baris_1004 + 1):
        print(" ", end="")
        print(angka_1004, end="")
    for spasi_1004 in range(2 * (n_1004 - baris_1004)):
        print(" ", end="")
    print(" |")

# ---------- Bingkai bawah ----------
print("#", end="")
for garis_1004 in range(4 * n_1004 + 5):
    print("=", end="")
print("#")
```

**Penjelasan :** Program tugas5_2611531004.py dirancang untuk mencetak pola Jam Pasir Kristal Palindromik Berbingkai secara dinamis, dengan ukuran yang ditentukan oleh nilai N dari pengguna. Seluruh pola (bingkai, spasi, dan deret angka) dibentuk hanya dengan perulangan for bersarang dan print(..., end=""), tanpa perkalian string. Alur programnya sebagai berikut:

1. **Input Ukuran Skala:** Program mencetak judul, lalu meminta nilai N melalui input() dan mengonversinya ke tipe integer dengan int(). Nilai disimpan pada variabel n_1004. Semua variabel berakhiran 4 digit terakhir NIM (1004) sesuai ketentuan.
2. **Bingkai Atas:** Program mencetak karakter pagar sudut (#), lalu perulangan for garis_1004 in range(4 * n_1004 + 5) mencetak karakter "=" satu per satu, dan ditutup dengan karakter # di ujung kanan. Jumlah 4N + 5 berasal dari lebar isi setiap baris terpanjang, yaitu 1 spasi padding + 2N karakter deret mundur + 3 karakter "<*>" + 2N karakter deret maju + 1 spasi padding.
3. **Fase 1 (Jam Pasir Atas):** Perulangan luar for baris_1004 in range(n_1004, 0, -1) menurunkan nilai baris dari N sampai 1. Pada setiap baris, perulangan dalam bekerja berurutan:
   - mencetak garis tegak (|) dan satu spasi padding,
   - for spasi_1004 in range(2 * (n_1004 - baris_1004)) mencetak spasi penyeimbang kiri,
   - for angka_1004 in range(baris_1004, 0, -1) mencetak deret mundur dari baris ke 1, masing-masing diikuti satu spasi,
   - mencetak poros kristal "<*>",
   - for angka_1004 in range(1, baris_1004 + 1) mencetak deret maju dari 1 ke baris, masing-masing diawali satu spasi,
   - mencetak spasi penyeimbang kanan (sebanyak 2 * (N - baris)), lalu satu spasi padding dan garis tegak (|) penutup.
4. **Fase 2 (Poros Titik Pusat):** Program mencetak garis tegak (|), lalu for spasi_1004 in range(2 * n_1004 + 1) untuk spasi kiri, karakter "<*>", spasi kanan dengan perulangan yang sama, dan garis tegak (|) penutup. Jumlah spasi 2N + 1 di tiap sisi membuat lebar baris ini tetap sama dengan baris lainnya.
5. **Fase 3 (Jam Pasir Bawah):** Struktur isinya sama dengan Fase 1, tetapi perulangan luar dibalik menjadi for baris_1004 in range(1, n_1004 + 1), sehingga deret angka dan poros kristal melebar kembali dari baris 1 sampai N dan membentuk bagian bawah jam pasir yang simetris dengan bagian atas.
6. **Bingkai Bawah:** Program mencetak ulang bingkai pembatas dengan cara yang sama seperti bingkai atas, yaitu karakter #, deret "=" sebanyak 4N + 5, lalu karakter #.
