# BAB II
# PEMBAHASAN

## 2.1 Dasar Teori

1. Python dan Fungsi Dasar
Python merupakan bahasa pemrograman tingkat tinggi yang banyak digunakan karena sintaksisnya yang ringkas. Fungsi print() digunakan untuk menampilkan output ke layar, sedangkan komentar (# untuk satu baris, """...""" untuk banyak baris) digunakan untuk mendokumentasikan kode agar lebih mudah dibaca.
2. Visual Studio Code (VS Code)
VS Code merupakan editor kode sumber terbuka (open-source) buatan Microsoft yang ringan namun kaya fitur, seperti IntelliSense, debugging terintegrasi, dan dukungan berbagai bahasa pemrograman melalui ekstensi. VS Code banyak direkomendasikan sebagai editor bagi mahasiswa pemula karena antarmukanya yang intuitif dan proses instalasinya yang mudah.
3. Git dan GitHub
Git adalah sistem version control terdistribusi yang digunakan untuk melacak perubahan pada kode program, sedangkan GitHub adalah platform daring yang menyimpan repository Git dan menjadi pusat kolaborasi bagi pengembang di seluruh dunia. Proses menghubungkan proyek lokal ke GitHub umumnya melibatkan tahapan inisialisasi repository (git init), penambahan perubahan (git add), pencatatan perubahan (git commit), serta pengunggahan ke repository jarak jauh (git push).

## 2.2 Kode Program

### 2.2.1 Program "NamaKelas_NIM"
Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan (tidak perlu screenshot terpisah untuk hasil run).

```python
# Program pertama
"""
Komentar
dua baris atau lebih
pertemuan 1 praktikum alpro
"""
print("Hello World")
print()
print("Departemen Informatika")
print("UNiversitas Andalas")
print("/ \\ // \\\\ /// \\\\\\")
print("program ini" + " dibuat oleh a\\n " + "Fabian Alhadi Zalinov dan 2611531004.")
```

**Penjelasan :**
1. Bagian pertama (baris komentar): Baris """Komentar dua baris atau lebih pertemuan 1 praktikum alpro""" merupakan komentar banyak baris (multi-line comment) yang diapit oleh tiga tanda kutip ("""..."""). Bagian ini tidak dieksekusi oleh Python dan berfungsi sebagai dokumentasi awal program, menjelaskan bahwa kode ini dibuat untuk pertemuan 1 praktikum Algoritma Pemrograman.
2. Bagian kedua (identitas program): Baris print("Hello World"), print(), print("Departemen Informatika"), dan print("UNiversitas Andalas") berfungsi menampilkan output identitas ke layar. print("Hello World") menampilkan teks sapaan standar sebagai program pertama, print() kosong digunakan untuk membuat baris kosong sebagai pemisah, sedangkan dua print() berikutnya menampilkan nama departemen dan universitas.
3. Bagian ketiga (dekorasi dan penggabungan string): Baris print("/ \\ // \\\\ /// \\\\\\") menampilkan karakter garis miring sebagai elemen dekoratif/hiasan pada output, dengan tanda \\ digunakan sebagai karakter escape untuk menampilkan tanda backslash (\) secara harfiah. Baris terakhir print("program ini" + " dibuat oleh a\\n " + "Fabian Alhadi Zalinov dan 2611531004.") menggabungkan (concatenate) tiga potongan string menggunakan operator + menjadi satu kalimat utuh, menampilkan keterangan pembuat program. Perlu dicatat bahwa \\n di sini ditulis dengan escape ganda sehingga akan tampil sebagai teks \n biasa, bukan sebagai baris baru (newline).
